# Problem 1: Ecploring Petal Length

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
from scipy import stats
```

```python
iris_url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'

iris = pd.read_csv(iris_url)

iris.head()
```

```python
iris.shape
iris["species"].value_counts()
```

```python
petal_length = iris['petal_length']

mean_petal_length = petal_length.mean()
std_petal_length = petal_length.std()

print('Mean petal length:', f'{mean_petal_length:.2f}')
print('Standard deviation of petal length:', f'{std_petal_length:.2f}')
```

```python
plt.figure(figsize=(8, 6))

plt.hist(petal_length, bins=20, density=True, alpha=0.7, color='lightblue', edgecolor='black')

plt.xlabel('Petal Length (cm)')
plt.ylabel('Density')
plt.title('Distribution of Petal Lengths in the Iris Dataset')

plt.show()
```
